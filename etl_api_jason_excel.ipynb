# ETL - extraindo e manipulando dados de API (IBGE)

### Consumir dados da API

```python
# para se conectar a URL do api do IBGE é necessário abrir biblioteca 'requests'

import requests

url = "https://servicodados.ibge.gov.br/api/v1/localidades/estados"

#fazer a requisição dos dados que estão na url

resposta = requests.get(url)

```

```python
# para saber se está conectado aos dados do API tem que aparecer 200

print(resposta.status_code)
```

```python
#criar variável que vai receber o conteúdo json -aquela formatação do site

estados = resposta.json()

print(estados [:3])   #vai pegar a primeira posição da lista até 3 - fatiamento
```

```python
#para visualizar as três linhas (mais apresentável)

print(estados [:3])
print("-" *30)

for estado in estados[:3]:
    print(estado)
```

### Gravando dados em arquivo JSON

```python
#converter as informações da variável estados em um arquivo json dentro do repositório atual

import json

with open("estados.json", "w", encoding="utf-8") as arquivo:
    json.dump(estados, arquivo, indent=4, ensure_ascii=False)


print("Arquivo estados.json criado com sucesso!")
```

```python
# apenas verificar se está ok (quantidade de estados)

qtde = 0     #vai iniciar em zero até o total

for estado in estados:   #para cada elemento, cada estado ele vai adicionar mais um
    qtde += 1
print(qtde)

```

#### Leitura do arquivo JSON



```python
import json

with open("estados.json", "r", encoding="utf-8") as arquivo:
    estados = json.load(arquivo)
```

### Converter JSON em DATAFRAME

```python
import pandas as pd

df_estados = pd.DataFrame(estados)

print(df_estados.head())   #exibe os cinco primeiros
```

### Selecionar colunas (apenas estado, sigla,região)

```python
#selecionar apenas as colunas desejadas

df_estados = df_estados[["sigla", "nome", "regiao",]]

#extrair o nome da região - acessar a coluna dentro do dicionário

df_estados["regiao"] = df_estados["regiao"].apply(lambda x: x["nome"])

df_estados.columns = ['UF', "Estado", "Regiao"]

print(df_estados)
```

```python
regioes = []
for região in df_estados["regiao"]:
    regioes.append(região("nome"))

    #poderia ter sido feita essa função ao invés da célula acima
```

### Gravar dados em Excel

```python
df_estados.to_excel("estados.xlsx")

df_estados.to_excel("estados.xlsx", index=False)  # com o index false não vai imprimir a lista de índice com números, apenas a partir de UF
print("Arquivo estados.xlsx criado com sucesso!")
```
